```python
# We import the necessary libraries to download and manipulate the data, and to visualize it.
import pandas as pd
import yfinance as yf
```

```python
# Removing the variables to the top of the notebook and use the config
from config import ASSETS, START_DATE, END_DATE, INTERVAL
```

```python
# Download Bitcoin data from yfinance from 2018-01-01 to actual date time, and interval of 1 day
btc_data = yf.download(
    start = START_DATE, 
    end = END_DATE, 
    tickers = ASSETS,
    interval = INTERVAL)
```

```python
# We check the data we downloaded
btc_data.info()
btc_data.describe()
btc_data.isnull().sum()
btc_data.head()
btc_data.tail()
```

```python
# Make it csv and save it in or raw data folder
btc_data.to_csv("../data/raw/btc_data_1d.csv")
```
