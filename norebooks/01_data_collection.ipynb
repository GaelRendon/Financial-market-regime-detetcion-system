```python
# We import the necessary libraries to download and manipulate the data, and to visualize it.
import pandas as pd
import yfinance as yf
import sys
import os

# Ruta absoluta a la raíz del proyecto
project_root = os.path.abspath("..")
if project_root not in sys.path:
    sys.path.append(project_root)
```

```python
# # Moveing the variables to the top of the notebook and made it consistent with the rest of the notebooks
# ASSET = "BTC-USD"
# START_DATE = "2018-01-01"
# END_DATE = pd.Timestamp.now()
# INTERVAL = "1d"
```

```python
# Removing the variables to the top of the notebook and use the config
from config import ASSETS, START_DATE, END_DATE, INTERVAL, RAW_DATA_PATH
```

```python
# Download, validate and save each asset

for asset in ASSETS:

    print(f"\n{'='*60}")
    print(f"Downloading {asset}")
    print(f"{'='*60}")

    data = yf.download(
        tickers=asset,
        start=START_DATE,
        end=END_DATE,
        interval=INTERVAL,
        auto_adjust=False
    )

    # Validation
    print("\nData Info:")
    data.info()

    print("\nMissing Values:")
    print(data.isnull().sum())

    print("\nFirst 5 Rows:")
    print(data.head())

    print("\nLast 5 Rows:")
    print(data.tail())

    # Dynamic filename
    asset_name = asset.lower().replace("-", "_")

    filename = f"{RAW_DATA_PATH}/{asset_name}_{INTERVAL}_raw.csv"

    # Let's remove the multiIndex problem we have
    if isinstance(data.columns, pd.MultiIndex):
        data.columns = data.columns.get_level_values(0)

    # Save CSV
    data.to_csv(filename)

    print(f"\nSaved: {filename}")
```
