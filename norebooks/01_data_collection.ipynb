```python
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
# Download Bitcoin data from yfinance from 2018-01-01 to actual date time, and interval of 1 day
btc_data = yf.download(start = "2018-01-01", end = pd.Timestamp.now(), tickers = "BTC-USD", interval = "1d")
```

```python
# Make it csv and save it in or raw data folder
df = btc_data.to_csv("../data/raw/btc_data_1d.csv")
```
